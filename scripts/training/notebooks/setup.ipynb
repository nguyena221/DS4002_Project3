This file sets up looseleaf for running the notebooks found in `DS4002_Project3/notebooks`. First it unzips the model file downloaded from `https://github.com/deepinsight/insightface/tree/master/model_zoo` (R100 (mxnet) MS1MV2), and clones both pytorch-insightface and our DS4002_Project3 repo. It then runs pytorch-insightface producing the pretrained iresnet100.pth. Next it copies this file into `DS4002_Project3/models/pretrained/`. Lastly it produces the SHA hash for use in the first notebook.

```python
# clean up repos from past runs
!rm -rf DS4002_Project3
!rm -rf pytorch-insightface

# unsip the model from the model zoo, clone gh repos
!unzip -l model-r100-ii.zip
!unzip -q -o model-r100-ii.zip
!git clone https://github.com/nizhib/pytorch-insightface.git
!git clone https://github.com/nguyena221/DS4002_Project3.git

# give pytorch-insightface the info it needs
%pip install mxnet==1.9.1
!cp model-r100-ii/model-0000.params pytorch-insightface/resource/iresnet100/model-0000.params
!cp model-r100-ii/model-symbol.json pytorch-insightface/resource/iresnet100/model-symbol.json

# magic mojo to run pytorch-insightface producing iresnet100.pth
!sed -i 's/from torchvision.models.utils import load_state_dict_from_url/from torch.hub import load_state_dict_from_url/' pytorch-insightface/insightface/iresnet.py
!cd pytorch-insightface && python -c "import numpy as np; np.bool = np.bool_; from scripts.convert import convert; convert('iresnet100')"

# copy over iresnet100.pth to our project
!cp -p pytorch-insightface/resource/iresnet100.pth DS4002_Project3/models/pretrained/iresnet100.pth
```

```python
# produce sha hash for use in next steps
!cd DS4002_Project3 && sha256sum models/pretrained/iresnet100.pth && du -h models/pretrained/iresnet100.pth
```
