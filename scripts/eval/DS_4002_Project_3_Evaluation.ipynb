```python
import hashlib
import json

root_seed = int(handoff["root_seed"])

def get_bootstrap_seed(root_seed, split):
    payload = json.dumps(
        {
            "root_seed": int(root_seed),
            "concern": "bootstrap",
            "scope": split
        },
        sort_keys=True,
        separators=(",", ":")
    ).encode("utf-8")

    return int.from_bytes(
        hashlib.sha256(payload).digest()[:4],
        "big"
    )

validation_seed = get_bootstrap_seed(root_seed, "validation")
test_seed = get_bootstrap_seed(root_seed, "test")

print("Root seed:", root_seed)
print("Validation bootstrap seed:", validation_seed)
print("Test bootstrap seed:", test_seed)
```

Implement the GPU version

```python
def gpu_weighted_triplet_accuracy(similarities, first_names, weights):
    """
    GPU implementation of the original weighted triplet accuracy.
    Uses strict cosine comparison: ties are incorrect.
    """
    correct = torch.zeros((), device="cuda", dtype=torch.float64)
    total = torch.zeros((), device="cuda", dtype=torch.float64)

    active = torch.where(weights > 0)[0]

    for anchor in active:
        positive = (
            (first_names == first_names[anchor])
            & (weights > 0)
        )
        positive[anchor] = False

        negative = (
            (first_names != first_names[anchor])
            & (weights > 0)
        )

        if not positive.any() or not negative.any():
            continue

        negative_sims = similarities[anchor, negative]
        negative_weights = weights[negative]

        order = torch.argsort(negative_sims, stable=True)
        sorted_sims = negative_sims[order]
        sorted_weights = negative_weights[order]

        cumulative_weights = torch.cumsum(sorted_weights, dim=0)

        positive_sims = similarities[anchor, positive]
        positions = torch.searchsorted(
            sorted_sims, positive_sims, right=False
        )

        lower_weights = torch.zeros_like(
            positions, dtype=torch.float64
        )

        valid = positions > 0
        lower_weights[valid] = cumulative_weights[
            positions[valid] - 1
        ].to(torch.float64)

        positive_weights = weights[positive].to(torch.float64)
        anchor_weight = weights[anchor].to(torch.float64)

        correct += anchor_weight * torch.sum(
            positive_weights * lower_weights
        )

        total += (
            anchor_weight
            * positive_weights.sum()
            * weights[negative].sum()
        )

    if total.item() == 0:
        raise ValueError("No valid triplets")

    return (correct / total).item()
```

Test the batched GPU version on a small sample

```python
import torch
import torch.nn.functional as F
import numpy as np

def gpu_batched_triplet_accuracy(similarities, labels, weights):
    """
    Exact weighted triplet accuracy for multiple bootstrap replicates.
    similarities: [N, N] GPU tensor
    labels:       [N] GPU tensor of integer class labels
    weights:      [R, N] GPU tensor of bootstrap multiplicities
    Returns:      [R] GPU tensor of accuracies
    """
    n = len(labels)
    device = similarities.device

    correct_total = torch.zeros(
        weights.shape[0], device=device, dtype=torch.int64
    )
    triplet_total = torch.zeros(
        weights.shape[0], device=device, dtype=torch.int64
    )

    anchor_block = 8

    for start in range(0, n, anchor_block):
        end = min(start + anchor_block, n)

        sim_block = similarities[start:end]
        label_block = labels[start:end]

        # Identify negatives; same-name images are not negatives.
        negative_mask = label_block[:, None] != labels[None, :]

        # Sort negative similarities. Non-negatives are placed at infinity.
        sorted_sims, sorted_idx = torch.sort(
            sim_block.masked_fill(~negative_mask, float("inf")),
            dim=1
        )

        valid_sorted = torch.gather(negative_mask, 1, sorted_idx)

        # Count negatives with similarity strictly below each candidate.
        lower_counts = torch.searchsorted(
            sorted_sims.contiguous(),
            sim_block.contiguous(),
            right=False
        )

        # Identify valid positive pairs, excluding each anchor itself.
        positive_mask = label_block[:, None] == labels[None, :]
        row_indices = torch.arange(start, end, device=device)
        positive_mask[
            torch.arange(end - start, device=device), row_indices
        ] = False

        for rep_start in range(0, weights.shape[0], 64):
            rep_end = min(rep_start + 64, weights.shape[0])
            w = weights[rep_start:rep_end]

            # Bootstrap weights in sorted-negative order.
            sorted_weights = w[:, sorted_idx]
            sorted_weights = sorted_weights * valid_sorted[None, :, :]

            cumulative = torch.cumsum(
                sorted_weights, dim=2, dtype=torch.int32
            )
            del sorted_weights

            positions = lower_counts.clamp(max=n - 1)
            positions = positions[None, :, :].expand(
                rep_end - rep_start, -1, -1
            )

            lower_weight = torch.gather(cumulative, 2, positions)
            del cumulative

            # A position of zero means no negative is strictly less similar.
            lower_weight = lower_weight * (lower_counts > 0)[None, :, :]

            positive_weights = (
                w[:, None, :] * positive_mask[None, :, :]
            )

            anchor_weights = w[:, start:end]

            correct = (
                positive_weights.to(torch.int64)
                * lower_weight.to(torch.int64)
            ).sum(dim=2)

            correct = (
                correct * anchor_weights.to(torch.int64)
            ).sum(dim=1)

            positive_sums = positive_weights.sum(dim=2).to(torch.int64)
            negative_sums = (
                w[:, None, :] * negative_mask[None, :, :]
            ).sum(dim=2).to(torch.int64)

            totals = (
                anchor_weights.to(torch.int64)
                * positive_sums
                * negative_sums
            ).sum(dim=1)

            correct_total[rep_start:rep_end] += correct
            triplet_total[rep_start:rep_end] += totals

    if torch.any(triplet_total == 0):
        raise ValueError("At least one replicate has no valid triplets.")

    return correct_total.double() / triplet_total.double()


# Test on the same 256 validation images
data = embedding_data["Baseline"]
ids = data["validation_h5_index"][:256]
embeddings = data["validation_embeddings"][:256]

labels = manifest.set_index("h5_index").loc[
    ids, "first_name"
].to_numpy()

weights = bootstrap_weights["validation"][:3, :256]

gpu_embeddings = torch.tensor(
    embeddings, dtype=torch.float32, device="cuda"
)
gpu_labels = torch.tensor(
    pd.factorize(labels)[0], dtype=torch.long, device="cuda"
)
gpu_weights = torch.tensor(
    weights, dtype=torch.int32, device="cuda"
)

normalized = F.normalize(gpu_embeddings, dim=1)
similarities = normalized @ normalized.T

gpu_results = gpu_batched_triplet_accuracy(
    similarities, gpu_labels, gpu_weights
).cpu().numpy()

cpu_results = np.array([
    weighted_triplet_accuracy(
        pairwise_cosine_similarity(embeddings),
        ids,
        labels,
        weights[i]
    )
    for i in range(3)
])

print("CPU results:", cpu_results)
print("Batched GPU results:", gpu_results)
print("Maximum difference:", np.max(np.abs(cpu_results - gpu_results)))
print("All match:", np.allclose(cpu_results, gpu_results, atol=1e-7))
```

Compare individual anchor contributions

```python
# Compare CPU and GPU weighted triplet counts for the first 8 anchors
cpu_sim = pairwise_cosine_similarity(embeddings)
w0 = weights[0].astype(np.int64)

for anchor in range(8):
    positive = (labels == labels[anchor]).copy()
    positive[anchor] = False
    positive &= w0 > 0
    negative = (labels != labels[anchor]) & (w0 > 0)

    if not positive.any() or not negative.any():
        continue

    neg_sims = cpu_sim[anchor, negative]
    neg_weights = w0[negative]
    order = np.argsort(neg_sims, kind="stable")
    sorted_sims = neg_sims[order]
    cumulative = np.cumsum(neg_weights[order])

    lower = np.searchsorted(
        sorted_sims, cpu_sim[anchor, positive], side="left"
    )
    weighted_lower = np.where(
        lower > 0, cumulative[np.maximum(lower - 1, 0)], 0
    )

    cpu_correct = (
        w0[anchor] * np.dot(w0[positive], weighted_lower)
    )
    cpu_total = (
        w0[anchor] * w0[positive].sum() * w0[negative].sum()
    )

    # GPU calculation for this same anchor
    gpu_neg_mask = gpu_labels != gpu_labels[anchor]
    gpu_pos_mask = gpu_labels == gpu_labels[anchor]
    gpu_pos_mask[anchor] = False

    gpu_neg_sims = similarities[anchor, gpu_neg_mask]
    gpu_neg_weights = gpu_weights[0, gpu_neg_mask]
    gpu_order = torch.argsort(gpu_neg_sims, stable=True)
    gpu_sorted_sims = gpu_neg_sims[gpu_order]
    gpu_cumulative = torch.cumsum(
        gpu_neg_weights[gpu_order], dim=0
    )

    gpu_lower = torch.searchsorted(
        gpu_sorted_sims,
        similarities[anchor, gpu_pos_mask],
        right=False
    )
    gpu_weighted_lower = torch.where(
        gpu_lower > 0,
        gpu_cumulative[(gpu_lower - 1).clamp(min=0)],
        0
    )

    gpu_correct = (
        gpu_weights[0, anchor]
        * (
            gpu_weights[0, gpu_pos_mask].long()
            * gpu_weighted_lower.long()
        ).sum()
    ).item()

    gpu_total = (
        gpu_weights[0, anchor]
        * gpu_weights[0, gpu_pos_mask].sum()
        * gpu_weights[0, gpu_neg_mask].sum()
    ).item()

    print(
        f"Anchor {anchor}: "
        f"CPU correct/total = {cpu_correct}/{cpu_total}; "
        f"GPU correct/total = {gpu_correct}/{gpu_total}"
    )
```

Replacing the batched function with this corrected version

```python
def gpu_batched_triplet_accuracy(similarities, labels, weights):
    n = len(labels)
    device = similarities.device

    correct_total = torch.zeros(
        weights.shape[0], device=device, dtype=torch.int64
    )
    triplet_total = torch.zeros(
        weights.shape[0], device=device, dtype=torch.int64
    )

    anchor_block = 8

    for start in range(0, n, anchor_block):
        end = min(start + anchor_block, n)

        sim_block = similarities[start:end]
        label_block = labels[start:end]

        negative_mask = label_block[:, None] != labels[None, :]

        sorted_sims, sorted_idx = torch.sort(
            sim_block.masked_fill(~negative_mask, float("inf")),
            dim=1
        )

        valid_sorted = torch.gather(negative_mask, 1, sorted_idx)

        lower_counts = torch.searchsorted(
            sorted_sims.contiguous(),
            sim_block.contiguous(),
            right=False
        )

        positive_mask = label_block[:, None] == labels[None, :]

        # Correctly exclude each anchor's own image.
        positive_mask[
            torch.arange(end - start, device=device),
            torch.arange(start, end, device=device)
        ] = False

        for rep_start in range(0, weights.shape[0], 64):
            rep_end = min(rep_start + 64, weights.shape[0])
            w = weights[rep_start:rep_end]

            sorted_weights = w[:, sorted_idx]
            sorted_weights = sorted_weights * valid_sorted[None, :, :]

            cumulative = torch.cumsum(
                sorted_weights, dim=2, dtype=torch.int32
            )

            positions = lower_counts.clamp(max=n - 1)
            positions = positions[None, :, :].expand(
                rep_end - rep_start, -1, -1
            )

            lower_weight = torch.gather(cumulative, 2, positions)
            lower_weight = lower_weight * (lower_counts > 0)[None, :, :]

            positive_weights = w[:, None, :] * positive_mask[None, :, :]
            anchor_weights = w[:, start:end]

            correct = (
                positive_weights.to(torch.int64)
                * lower_weight.to(torch.int64)
            ).sum(dim=2)

            correct = (
                correct * anchor_weights.to(torch.int64)
            ).sum(dim=1)

            positive_sums = positive_weights.sum(dim=2).to(torch.int64)
            negative_sums = (
                w[:, None, :] * negative_mask[None, :, :]
            ).sum(dim=2).to(torch.int64)

            totals = (
                anchor_weights.to(torch.int64)
                * positive_sums
                * negative_sums
            ).sum(dim=1)

            correct_total[rep_start:rep_end] += correct
            triplet_total[rep_start:rep_end] += totals

    if torch.any(triplet_total == 0):
        raise ValueError("At least one replicate has no valid triplets.")

    return correct_total.double() / triplet_total.double()


# Retest all three bootstrap replicates
gpu_results = gpu_batched_triplet_accuracy(
    similarities, gpu_labels, gpu_weights
).cpu().numpy()

print("CPU results:", cpu_results)
print("Corrected batched GPU results:", gpu_results)
print("Maximum difference:", np.max(np.abs(cpu_results - gpu_results)))
print("All match:", np.allclose(cpu_results, gpu_results, atol=1e-7))
```

Replacing the batched function and tests all three replicates against the CPU results

```python
def gpu_batched_triplet_accuracy(similarities, labels, weights):
    n = len(labels)
    device = similarities.device
    n_reps = weights.shape[0]

    correct_total = torch.zeros(
        n_reps, device=device, dtype=torch.float64
    )
    triplet_total = torch.zeros(
        n_reps, device=device, dtype=torch.float64
    )

    for anchor in range(n):
        positive_idx = torch.where(labels == labels[anchor])[0]
        positive_idx = positive_idx[positive_idx != anchor]

        negative_idx = torch.where(labels != labels[anchor])[0]

        if len(positive_idx) == 0 or len(negative_idx) == 0:
            continue

        negative_sims = similarities[anchor, negative_idx]
        sorted_sims, order = torch.sort(negative_sims, stable=True)
        sorted_negative_idx = negative_idx[order]

        positive_sims = similarities[anchor, positive_idx]

        lower_counts = torch.searchsorted(
            sorted_sims, positive_sims, right=False
        )

        for rep_start in range(0, n_reps, 64):
            rep_end = min(rep_start + 64, n_reps)
            w = weights[rep_start:rep_end]

            negative_weights = w[:, sorted_negative_idx]
            cumulative = torch.cumsum(negative_weights, dim=1)

            positions = (lower_counts - 1).clamp(min=0)
            lower_weights = cumulative[:, positions]

            lower_weights = lower_weights * (
                lower_counts > 0
            )[None, :]

            positive_weights = w[:, positive_idx]
            anchor_weights = w[:, anchor]

            correct = (
                positive_weights.to(torch.float64)
                * lower_weights.to(torch.float64)
            ).sum(dim=1) * anchor_weights

            total = (
                positive_weights.sum(dim=1).to(torch.float64)
                * w[:, negative_idx].sum(dim=1).to(torch.float64)
                * anchor_weights.to(torch.float64)
            )

            correct_total[rep_start:rep_end] += correct
            triplet_total[rep_start:rep_end] += total

    if torch.any(triplet_total == 0):
        raise ValueError("At least one replicate has no valid triplets.")

    return correct_total / triplet_total


# Test against the original CPU implementation
gpu_results = gpu_batched_triplet_accuracy(
    similarities, gpu_labels, gpu_weights
).cpu().numpy()

print("CPU results:", cpu_results)
print("GPU results:", gpu_results)
print("Maximum difference:", np.max(np.abs(cpu_results - gpu_results)))
print("All match:", np.allclose(cpu_results, gpu_results, atol=1e-7))
```

```python
import os
import numpy as np
import pandas as pd

models = ["Baseline", "Triplet Margin", "ArcFace", "SupCon"]
splits = ["validation", "test"]

# Load all saved bootstrap results from Google Drive
bootstrap_results = {}

for split in splits:
    for model in models:
        filename = f"{model.lower().replace(' ', '_')}_{split}_bootstrap.npz"
        filepath = os.path.join(results_folder, filename)

        data = np.load(filepath)
        values = data["accuracies"]

        assert len(values) == 2000, f"Incorrect replicate count: {filepath}"

        bootstrap_results[(model, split)] = values

print("All eight bootstrap files loaded successfully.")

# Calculate observed full-dataset accuracy for each model and split
observed_accuracy = {}

for split in splits:
    for model in models:
        data = embedding_data[model]
        ids = data[f"{split}_h5_index"]
        embeddings = data[f"{split}_embeddings"]

        labels = manifest.set_index("h5_index").loc[
            ids, "first_name"
        ].to_numpy()

        observed_accuracy[(model, split)] = blocked_triplet_accuracy(
            embeddings, ids, labels
        )

# Calculate paired lifts and percentile confidence intervals
summary_rows = []
paired_differences = {}

for split in splits:
    baseline = bootstrap_results[("Baseline", split)]

    for model in models:
        if model == "Baseline":
            continue

        comparison = bootstrap_results[(model, split)]

        # Same replicate number = same bootstrap sample
        differences = (comparison - baseline) * 100
        paired_differences[(model, split)] = differences

        ci_lower, ci_upper = np.percentile(
            differences, [2.5, 97.5]
        )

        observed_lift = (
            observed_accuracy[(model, split)]
            - observed_accuracy[("Baseline", split)]
        ) * 100

        summary_rows.append({
            "Split": split.capitalize(),
            "Model": model,
            "Baseline Accuracy (%)": (
                observed_accuracy[("Baseline", split)] * 100
            ),
            "Model Accuracy (%)": (
                observed_accuracy[(model, split)] * 100
            ),
            "Observed Lift (pp)": observed_lift,
            "Bootstrap Mean Lift (pp)": differences.mean(),
            "95% CI Lower (pp)": ci_lower,
            "95% CI Upper (pp)": ci_upper
        })

results_table = pd.DataFrame(summary_rows)

# Display results
display(
    results_table.round(4)
)

# Save the summary and paired bootstrap distributions
results_table.to_csv(
    os.path.join(results_folder, "paired_bootstrap_summary.csv"),
    index=False
)

np.savez_compressed(
    os.path.join(results_folder, "paired_bootstrap_differences.npz"),
    **{
        f"{model.lower().replace(' ', '_')}_{split}": values
        for (model, split), values in paired_differences.items()
    }
)

print("\nSummary and paired bootstrap differences saved to Google Drive.")
```

Run the checksum audit

```python
import hashlib
from pathlib import Path

# Paths
root = Path("/content/DS4002_Project3")
package = root / "artifacts/face_name_metric_learning"

# Expected checksums from handoff_manifest.json
expected_checksums = handoff["checksums"]

# Map handoff paths to their actual locations
file_paths = {
    "embeddings/arcface_selected.npz":
        package / "embeddings/arcface_selected.npz",

    "embeddings/baseline_iresnet100.npz":
        package / "embeddings/baseline_iresnet100.npz",

    "embeddings/supcon_selected.npz":
        package / "embeddings/supcon_selected.npz",

    "embeddings/triplet_margin_selected.npz":
        package / "embeddings/triplet_margin_selected.npz",

    "manifest.csv":
        root / "data/processed/manifest.csv",

    "trial_results.csv":
        package / "trial_results.csv",
}

def sha256_file(path):
    digest = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()

print("CHECKSUM AUDIT")
print("=" * 60)

passed = []
failed = []

for handoff_name, path in file_paths.items():
    if handoff_name not in expected_checksums:
        print(f"MISSING EXPECTED CHECKSUM: {handoff_name}")
        failed.append(handoff_name)
        continue

    if not path.exists():
        print(f"FILE NOT FOUND: {handoff_name}")
        failed.append(handoff_name)
        continue

    actual = sha256_file(path)
    expected = expected_checksums[handoff_name]

    if actual == expected:
        print(f"PASS: {handoff_name}")
        passed.append(handoff_name)
    else:
        print(f"FAIL: {handoff_name}")
        print("  Expected:", expected)
        print("  Actual:  ", actual)
        failed.append(handoff_name)

print("\nSUMMARY")
print(f"Passed: {len(passed)}")
print(f"Failed: {len(failed)}")
print("faces.h5: NOT CHECKED (intentionally not uploaded)")

assert len(failed) == 0, "One or more checksum checks failed."
```
